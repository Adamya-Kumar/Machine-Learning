```python
import numpy as np
import matplotlib.pyplot as pd
```

```python
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])
w_init=np.array([0.0,0.0])
b_init=0.0
```

```python
def sigmod_function(z):
    g=(1/(1+np.exp(-z)))
    return g
```

```python
def cost_function(w,b,x,y):
    m=x.shape[0]
    cost=0
    for i in range(m):
        z_i = np.dot(w,x[i])+b
        f_wb_i = sigmod_function(z_i)
        j_wb = -y[i]*np.log(f_wb_i) - (1-y[i])*np.log(1-f_wb_i)
        cost = cost + j_wb
    cost = cost / m
    return cost
        
```

```python
def derivate_grdient_descent(w,b,x,y):
    m,n=x.shape
    dj_dw=np.zeros((n,))
    dj_db=0
    for i in range(m):
        z_i = np.dot(w,x[i])+b
        f_wb_i = sigmod_function(z_i)
        error_i = f_wb_i - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + error_i * x[i,j]
        dj_db = dj_db + error_i
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    return dj_dw,dj_db
```

```python
def gradient_descent_function(w,b,itr,alpha,x,y):
    m,n=x.shape
    J_History=[]
    P_History=[]
    dj_dw=np.zeros((n,))
    dj_db=0
    limit=100
    count=0
    for i in range(itr):
        dj_dw,dj_db = derivate_grdient_descent(w,b,x,y)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        
        if i < itr:
            cost = cost_function(w,b,x,y)
            J_History.append(cost)
            P_History.append((w,b))
        if limit == count:
            print(f"cost:{cost}, w:{w}, b:{b}, dj_dw:{dj_dw}, dj_db:{dj_db}\n")
            count=0
        elif limit > count:
            count = count + 1
        
    return w,b,J_History,P_History
            
```

```python
w,b,J_History,P_History =  gradient_descent_function(w_init,b_init,10000,0.1,X_train,y_train)
w_tmp  = np.zeros_like(X_train[0])
b_tmp  = 0.
w,b,J_History,P_History =  gradient_descent_function(w_tmp,b_tmp,10000,0.1,X_train,y_train)
print(w,b)
```
