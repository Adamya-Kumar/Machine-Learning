```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
```

```python
def compute_cost_function_linear_regression_with_regualarizated(w,b,x,y,lambda_=1):
    m,n=x.shape
    cost=0
    penltie=0
    for i in range(m):
        fwb_i = (np.dot(w,x[i])+b)
        error_i = (fwb_i - y[i])**2
        cost = cost + error_i
    cost = (cost / (2*m))
    
    for j in range(n):
        penltie = penltie + (w[j]**2)
    penltie = (lambda_/(2*m)) * penltie
    
    total_cost = cost + penltie
    return total_cost
    
```

```python
cost = compute_cost_function_linear_regression_with_regualarizated(w_tmp,b_tmp,X_tmp,y_tmp,8)
```

```python
def compute_derivate_function_linear_regression_with_regualarizated(w,b,x,y,lambda_=1):
    m,n=x.shape
    dj_dw=np.zeros((n,))
    dj_db=0
    for i in range(m):
        fwb_i = np.dot(w,x[i])+b
        fwb_i = fwb_i - y[i]
        for j in range(n):
            dj_dw[j] += fwb_i * x[i,j]
        dj_db = dj_db + fwb_i
    dj_dw = dj_dw / m
    for j in range(n):
        dj_dw[j] += (lambda_/m) * w[j]
    dj_db = dj_db / m
    return dj_dw,dj_db
```

```python
dj_dw, dj_db = compute_derivate_function_linear_regression_with_regualarizated(w_tmp, b_tmp, X_tmp, y_tmp, 0.7)
```

```python
print(dj_dw,dj_db)
```

```python
def gradient_descent(w, b, X, y, itr, alpha, lambda_):
    
    w = np.array(w)
    cost_history = []
    
    for i in range(itr):
        
        dj_dw, dj_db = compute_derivate_function_linear_regression_with_regualarizated(
            w, b, X, y, lambda_
        )
        
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        
        if i % 100 == 0:
            cost = compute_cost_function_linear_regression_with_regualarizated(
                w, b, X, y, lambda_
            )
            cost_history.append(cost)
            print(f"Iteration {i}: Cost = {cost:.4f}")
            
            if np.isnan(cost):
                print("Diverging! Reduce alpha.")
                break
    
    return w, b, cost_history
```
