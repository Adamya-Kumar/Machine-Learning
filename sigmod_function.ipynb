```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x_train=np.arange(-10,11)
x_train[4]=-10
print(x_train)
```

```python
def sigmod_function(z):
    g=(1/(1+np.exp(-z)))
    return g
```

```python
y_hat=sigmod_function(x_train)
print(y_hat)
```

```python
# Plot z vs sigmoid(z)
fig,ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(x_train, y_hat, c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
```
