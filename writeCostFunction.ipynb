```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train=np.array([1.,2.,3.,4.,5.])
y_train=np.array([100.0,200.0,300.0,400.0,500.0])
```

```python
print(x_train,y_train)
```

```python
def cost_function(w,b,x,y):
    m=x.shape[0]
    cost_sum=0
    for i in range(m):
        fwb_xi=(w*x[i])+b
        cost_i=(fwb_xi-y[i])**2
        cost_sum=cost_sum+cost_i
    total_cost=(1/(2*m))*cost_sum
    return total_cost
```

```python
j_wb=cost_function(50,20,x_train,y_train)
print(j_wb)
```

```python
def cost_function_try(w,b,x,y):
    total_cost=0
    m=x.shape[0]
    for i in range(m):
        fwb_i=(w*x[i])+b
        error_i = (fwb_i - y[i])**2
        total_cost = total_cost + error_i
    total_cost = (1/(2*m)) * total_cost
    return total_cost
    
```

```python
cost_function_try(50,20,x_train,y_train)
```
