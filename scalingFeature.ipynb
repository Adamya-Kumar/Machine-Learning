# cost function

```python
def compute_cost(w,b,X,y):
    m,n=X.shape
    cost=0.0
    for i in range(m):
        f_wb_i = np.dot(w,X[i]) + b
        cost = cost + (f_wb_i-y[i])**2
    cost = (cost * (1/(2*m)))
    return cost
```

# gradient descent



```python
def compute_derviate_function(w,b,X,y):
    cost=0.0
    m,n=X.shape
    dj_dw=np.zeros((n,))
    dj_db=0.0
    for i in range(m):
        error = (np.dot(X[i],w)+b) - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + error * X[i,j]
        dj_db = dj_db + error
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    return dj_dw,dj_db    
```

```python
compute_derviate_function(w,b,x_train,y_train)
```

```python
def compute_gradient_descent(w,b,X,y,itr,alpha):
    J_history=[]
    P_history=[]
    count=0
    limit=100
    for i in range(itr):
        dj_dw,dj_db = compute_derviate_function(w,b,X,y)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        if i < itr:
            J_history.append(compute_cost(w,b,X,y))
            P_history.append((w,b))
        if limit == count:
            print(f"cost:{J_history[-1]}, w: {w}, b:{b},dj_dw:{dj_dw},dj_db:{dj_db}")
            count = 0
        count = count + 1
    return w,b,J_history,P_history
    
```

```python
w,b,J_History,P_History = compute_gradient_descent(np.zeros((x_train[0].shape)),0,x_train,y_train,10000,0.05)
```

```python
def zscore_normalize_features(X):
    mu = np.mean(X,axis=0)
    std = np.std(X,axis=0)
    X_scaled = (X - mu) / std
    return X_scaled,mu,std
```
