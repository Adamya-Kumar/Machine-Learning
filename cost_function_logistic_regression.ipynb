```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
data = np.array([
    [0.5, 1.5, 0.],
    [1.0, 1.0, 0.],
    [1.5, 0.5, 0.],
    [3.0, 0.5, 1.],
    [2.0, 2.0, 1.],
    [1.0, 2.5, 1.]
])
```

```python
x_train=data[:,:2]
y_train=data[:,2]
```

```python
def sigmod_function(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
def compute_cost_function(x,y,w,b):
    m=x.shape[0]
    cost=0.0
    for i in range(m):
        f_wb_i = np.dot(w,x[i])+b
        fz_i=sigmod_function(f_wb_i)
        cost += -y[i]*np.log(fz_i) - (1-y[i])*np.log(1-fz_i)
    cost = cost / m
    return cost
```

```python
w_tmp = np.array([1,1])
b_tmp = -3
cost = compute_cost_function(x_train,y_train,w_tmp,b_tmp)
print(cost)
b_tmp=-4
cost = compute_cost_function(x_train,y_train,w_tmp,b_tmp)
print(cost)
```
